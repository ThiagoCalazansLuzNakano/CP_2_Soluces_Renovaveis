## 4. Matriz de correlação

Calcule a matriz de correlação das variáveis numéricas e crie um mapa de calor (*heatmap*). Em seguida, identifique as **cinco variáveis com maior correlação absoluta com `stab`**, sem considerar a própria coluna `stab`.

Considere também correlações negativas: quanto mais próximo de +1 ou -1, maior é a intensidade da relação linear.

```python
# Plot do Mapa de Calor (Heatmap) da matriz de correlação completa
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Matriz de Correlação")
plt.show()

correlacoes_com_stab = df.corr(numeric_only=True)['stab'].drop('stab')

# Identifica as 5 variáveis com maior correlação absoluta
cinco_maiores_correlacoes = correlacoes_com_stab.abs().nlargest(5)

# Exibe os nomes e valores das 5 variáveis
print("As 5 variáveis com maior correlação absoluta com 'stab':")
print(correlacoes_com_stab[cinco_maiores_correlacoes.index])
```

```python
# Cole, revise e execute aqui o código gerado com o apoio do Gemini
cinco_variaveis = []
for var, abs_corr in cinco_maiores_correlacoes.items():
  original_corr = correlacoes_com_stab[var]
  print(f"Variável: {var}, Correlação Original: {original_corr:.4f}")
  cinco_variaveis.append(var)

print(f"\nLista 'cinco_variaveis': {cinco_variaveis}")
```

```python
# DESCOMENTAR APÓS TERMINAR
# Criar os graficos de dispersão
plt.figure(figsize=(15, 10))
for i, var in enumerate(cinco_variaveis):
  plt.subplot(2, 3, i + 1) #2 linhas, 3 colunas
  sns.scatterplot( x=df[var], y=df['stab'])
  plt.title(f'Dispersão de {var} vs stab')
  plt.xlabel(var)
  plt.ylabel('stab')

plt.tight_layout()
plt.show()
```
